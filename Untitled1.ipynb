```python
def convert(obj):
    L = []
    for i in ast.literal_eval(obj):
        L.append(i['name'])
    return L
```

```python
import ast
```

```python
movies['genres'] = movies['genres'].apply(convert)
movies.head()
```

```python
movies['keywords'] = movies['keywords'].apply(convert)
```

```python
def covert3(obj):
    L = []
    counter = 0
    for i in ast.literal_eval(obj):
        if counter != 3:
            L.append(i['name'])
            counter+=1
        else:
          break
    return L
```

```python
movies['cast'] = movies['cast'].apply(covert3)
movies.head()
```

```python
movies['crew'][0]
```

```python
def fetch_director(text):
    L = []
    for i in ast.literal_eval(text):
        if i['job'] == 'Director':
            L.append(i['name'])
    return L
movies['crew'] = movies['crew'].apply(fetch_director)
```

```python
movies.sample(5)
```

```python
movies['overview'] = movies['overview'].apply(lambda x:x.split())
```

```python
movies.head()
```

```python
movies['genres'] = movies['genres'].apply(lambda x :[i.replace(" "," ")for i in x]).apply(lambda x :[i.replace(" "," ")for i in x])
movies['keywords'] = movies['keywords'].apply(lambda x :[i.replace(" "," ")for i in x])
movies['cast'] = movies['cast'].apply(lambda x :[i.replace(" "," ")for i in x])
movies['crew'] = movies['crew'].apply(lambda x :[i.replace(" "," ")for i in x])
```

```python
movies.head()
```

```python
movies['tags'] = movies['overview'] + movies['genres'] + movies['keywords'] + movies['cast'] + movies['crew']
```

```python
movies.head()
```

```python
nef_df = movies[['movie_id','title','tags']]
nef_df
```

```python
nef_df['tags']= nef_df['tags'].apply(lambda x: " ".join(x))
nef_df.head()
```

```python
import nltk
```

```python
from nltk.stem.porter import PorterStemmer
ps =  PorterStemmer()
```

```python
def stem(text):
  y = []

  for i in text.split():
      y.append(ps.stem(i))
  return " ".join(y)
```

```python
nef_df['tags'] = nef_df['tags'].apply(stem)
```

```python
from sklearn.feature_extraction.text import CountVectorizer
cv = CountVectorizer(max_features=5000,stop_words='english')

vector = cv.fit_transform(nef_df['tags']).toarray()
```

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
similarity = cosine_similarity(vector)
```

```python
def recommend(movie):
    index = nef_df[nef_df['title'] == movie].index[0]
    distances = sorted(list(enumerate(similarity[index])),reverse=True,key = lambda x: x[1])
    for i in distances[1:6]:
        print(nef_df.iloc[i[0]].title)
```

```python
recommend('Spectre')
```
